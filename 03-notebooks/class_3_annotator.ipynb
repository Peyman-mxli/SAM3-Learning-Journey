```python
!pip install supervision ultralytics

import cv2
from ultralytics import YOLO
import supervision as sv
from supervision.assets import download_assets, VideoAssets

download_assets(VideoAssets.PEOPLE_WALKING)

model = YOLO("yolov8n.pt")
```

```python
cap = cv2.VideoCapture('people-walking.mp4')
ret, first_frame = cap.read()
cap.release()

results = model(first_frame)[0]
detections = sv.Detections.from_ultralytics(results)
```

## Bounding Box

```python
box_annotator = sv.BoxAnnotator()
label_annotator = sv.LabelAnnotator()
labels = [
    f"{results.names[c]} {conf:.0%}"
    for c, conf in zip(detections.class_id, detections.confidence)
]
annotated = box_annotator.annotate(first_frame.copy(), detections = detections)
annotated = label_annotator.annotate(annotated, detections = detections, labels = labels)
sv.plot_image(annotated)
```

## Round Box

```python
round_box_annotator = sv.RoundBoxAnnotator()
annotated = round_box_annotator.annotate(first_frame.copy(), detections = detections)
sv.plot_image(annotated)
```

## Box Corner Annotator

```python
corner_box_annotator = sv.BoxCornerAnnotator()
annotated = corner_box_annotator.annotate(first_frame.copy(), detections = detections)
sv.plot_image(annotated)
```

## Oriented Box Annotator

Important obb model works for detecting objects in aerial view

```python
image = cv2.imread('original.jpg')
oriented_model = YOLO('yolov8s-obb.pt')
oriented_results = oriented_model(image)[0]
oriented_detections = sv.Detections.from_ultralytics(oriented_results)
print(len(oriented_results))
oriented_box_annotator = sv.OrientedBoxAnnotator()
annotated = oriented_box_annotator.annotate(image.copy(), detections = oriented_detections)
sv.plot_image(annotated)
```
